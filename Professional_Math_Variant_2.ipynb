Задание №18

```python
import math

x = 4.52
eps = 1e-6
pi = 3.141592653589793

# приводим угол к удобному промежутку
y = x - pi

# первый член ряда
term = y
s = term
count = 1
k = 0

while True:
    next_term = -term * y * y / ((2 * k + 2) * (2 * k + 3))

    if abs(next_term) < eps:
        break

    s = s + next_term
    term = next_term
    k = k + 1
    count = count + 1

# sin(4.52) = -sin(y)
result = -s

# проверка
check = math.sin(x)
error = abs(result - check)

print("Приведенный угол:", y)
print("sin(4.52) =", round(result, 6))
print("Число членов:", count)
print("Следующий член:", next_term)
print("Погрешность:", error)
```

Задание №19

```python
import math

x = 0.32
eps = 1e-8

term = 1.0
s = term
count = 1
k = 0

while True:
    next_term = -term * x * x / ((2 * k + 1) * (2 * k + 2))

    if abs(next_term) < eps:
        break

    s = s + next_term
    term = next_term
    k = k + 1
    count = count + 1

# проверка библиотечной функцией
check = math.cos(x)
error = abs(s - check)

print("cos(0.32) =", round(s, 6))
print("Число членов:", count)
print("Следующий член:", next_term)
print("Погрешность:", error)
```

Задание №20

```python
import math

t = 0.35
eps = 1e-8

# ряд для e^t
term = 1.0
s_exp = term
count_exp = 1
k = 0

while True:
    next_term = term * t / (k + 1)

    if abs(next_term) < eps:
        break

    s_exp = s_exp + next_term
    term = next_term
    k = k + 1
    count_exp = count_exp + 1


# ряд для cos(t)
term = 1.0
s_cos = term
count_cos = 1
k = 0

while True:
    next_term = -term * t * t / ((2 * k + 1) * (2 * k + 2))

    if abs(next_term) < eps:
        break

    s_cos = s_cos + next_term
    term = next_term
    k = k + 1
    count_cos = count_cos + 1


# значение F(t)
F = s_exp * s_cos

# проверка библиотечными функциями
check = math.exp(t) * math.cos(t)
error = abs(F - check)

print("e^t =", round(s_exp, 6))
print("cos(t) =", round(s_cos, 6))
print("F(t) =", round(F, 6))
print("Число членов ряда e^t:", count_exp)
print("Число членов ряда cos(t):", count_cos)
print("Погрешность:", error)
```
